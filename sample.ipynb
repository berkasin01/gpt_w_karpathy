```python
import os, urllib.request

os.makedirs("data", exist_ok=True)
url = "https://raw.githubusercontent.com/karpathy/char-rnn/master/data/tinyshakespeare/input.txt"
urllib.request.urlretrieve(url, "data/input.txt")

with open("data/input.txt", "r", encoding="utf-8") as f:
    text = f.read()

print(len(text))
print(text[:300])
```

```python
len(text)
```

```python
chars = sorted(list(set(text))) ##gets all the unique characters in the text, puts them in a list, ordered
vocab_size = len(chars) ##how many unique characters
print(vocab_size)
print("".join(chars))
```

```python
stoi = { ch:i for i,ch in enumerate(chars) } ##builds a dictionary of chars
itos = { i:ch for i,ch in enumerate(chars) } ##builds a dictionary of chars in reverse
```

```python
encode = lambda s: [stoi[c] for c in s] ## encoder: take a string, output a list of integers
decode = lambda l: "".join([itos[i] for i in l]) ##decoder: take a list of integers, output string
```

```python
print(encode("hiiii there"))
print(decode(encode("hiii there")))
```
